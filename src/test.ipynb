```python
from load_data import DataSet

dir_base = '/home/lalonso/PhD/VisHD/data/'
dataset = DataSet(dir_base=dir_base, img_file="Visium_HD_Mouse_Small_Intestine_tissue_image.btf", bins_size=8)
img, adata, df_tissue_positions = dataset.load()
print("Data loaded succesfully")
```

```python
from utils.bins2spots import Bins2Spots

b2s = Bins2Spots(adata, df_tissue_positions, bins_size=8)
```

```python
from sample import Bins, Spots

spots = Spots(b2s.spot_polygons, adata=b2s.convert(normalize=False))
bins = Bins(b2s.bin_points, adata=adata, on='index')
```

```python
gene1 = list(adata.var_names)[0]
gene2 = list(adata.var_names)[2]
```

```python
import sys
sys.path.append("./..")
from bokeh_server.loader import SourceManager
```

```python
size = 5000
bbox = [0,size,0,size]
manager = SourceManager(spots, [gene1, gene2])
```

```python
import matplotlib.pyplot as plt

def plot(source):
    plt.scatter(x=source.data['x'], y=source.data['y'], c=source.data[manager.colnames[0]])
    plt.show()
```

```python
manager.add_data(bbox=bbox)
plot(manager.source)
manager.add_data(bbox=[b+size for b in bbox])
plot(manager.source)
manager.remove_data(bbox=[b+size/2 for b in bbox])
plot(manager.source)
manager.add_data(bbox=bbox)
plot(manager.source)
manager.remove_further_data(0,0, distance=10000)
plot(manager.source)
```
